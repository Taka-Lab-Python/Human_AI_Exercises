rinna/japanese-gpt2-medium, 次にくるトークンを予測するだけのモデル

学習済みLLMを使って「次のトークンの候補と確率を表示する」

日本語の文章生成モデル rinna/japanese-gpt2-medium を使う

```python
# 【コード01】

!pip -q install transformers sentencepiece

import torch
from transformers import AutoTokenizer, AutoModelForCausalLM
from transformers import logging

logging.set_verbosity_error()
print("1分ほどお待ちください")

model_name = "rinna/japanese-gpt2-medium"

# そのモデルに対応するTokenizerを持ってくる
tokenizer = AutoTokenizer.from_pretrained(
    model_name,
    use_fast=False
)

# Hugging Face上のモデルを取得し、Colab上で使える状態に読み込む
model = AutoModelForCausalLM.from_pretrained(model_name)

# 学習はしない。この学習済みモデルを使って予測だけする
model.eval()

print("モデルの設定が完了しました。")
```

```python
# 【コード02】

text = "日本の首都は"

inputs = tokenizer(text, return_tensors="pt", add_special_tokens=True)

# IDをトークンに変換
tokens = tokenizer.convert_ids_to_tokens(inputs["input_ids"][0])
print("tokens：", tokens, "\n")

with torch.no_grad():
    outputs = model(**inputs)

scores = outputs.logits[0, -1]

#スコアを確率に変換する
probabilities = torch.softmax(scores, dim=-1)

top_probs, top_ids = torch.topk(probabilities, 7)

for p, token_id in zip(top_probs, top_ids):
    token = tokenizer.decode([token_id.item()])
    print(token, f"{p.item()*100:.2f}%")
```
